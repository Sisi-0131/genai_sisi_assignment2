```python
# q1: Given an input image of size 32×32×3 
# and a convolutional layer with 8 filters of size 5×5, stride 1, and no padding, what is the output size?

input_size = 32
kernel_size = 5
stride = 1
padding = 0
number_of_filters = 8

output_size = (
    input_size + 2 * padding - kernel_size
) // stride + 1

print("Output height:", output_size)
print("Output width:", output_size)
print("Output channels:", number_of_filters)
print(
    "Answer:",
    output_size,
    "x",
    output_size,
    "x",
    number_of_filters,
)
```

```python
# q2: How does the output size change if padding is changed to “same”?

# the input is 32 × 32, so the output spatial size is also 32 × 32.
# The layer still has 8 filters, so it produces 8 output channels.
# Answer is 32 × 32 × 8
```

```python
# q3: If you apply a 3×3 filter with stride 2 and no padding to a 64×64 input, what is the output spatial size?
input_size = 64
kernel_size = 3
stride = 2
padding = 0

output_size = (
    input_size + 2 * padding - kernel_size
) // stride + 1

print("Output height:", output_size)
print("Output width:", output_size)
print("Answer:", output_size, "x", output_size)
```

```python
# q4: You apply a max-pooling layer of size 2×2 with stride 2 on a 16×16 feature map. What is the output size?
input_size = 16
pooling_size = 2
stride = 2
padding = 0

output_size = (
    input_size + 2 * padding - pooling_size
) // stride + 1

print("Output height:", output_size)
print("Output width:", output_size)
print("Answer:", output_size, "x", output_size)
```
